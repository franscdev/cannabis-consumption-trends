### 3.1 Finding 1 — Historical Cannabis Prevalence by Educational Grade

**Research Question**

How has past-year cannabis prevalence evolved across educational grades between 1975 and 2025?

**Key Observations**

The Monitoring the Future dataset reveals substantial differences in cannabis prevalence across educational grades.

The historical coverage differs by grade:

- **12th Grade:** 1975–2025.
- **10th Grade:** 1991–2025.
- **8th Grade:** 1991–2025.

This distinction is important because the full historical series cannot be compared directly across all three grades before 1991.

**Historical Prevalence Peaks**

| Educational Grade | Highest Observed Prevalence | Year |
|---|---:|---:|
| 8th Grade | 18.3% | 1996 |
| 10th Grade | 34.8% | 1997 |
| 12th Grade | 50.8% | 1979 |

These values represent the highest observed past-year cannabis prevalence estimates within each grade's available historical series.

**1991–2025 Comparison**

| Educational Grade | 1991 | 2025 | Change |
|---|---:|---:|---:|
| 8th Grade | 6.2% | 7.6% | +1.4 pp |
| 10th Grade | 16.5% | 15.6% | −0.9 pp |
| 12th Grade | 23.9% | 25.7% | +1.8 pp |

**Interpretation**

Cannabis prevalence has not followed a consistent upward or downward trajectory across the entire historical period.

Instead, the data reveal distinct periods of increasing and decreasing prevalence.

The 2025 estimates remain substantially below the historical peaks observed in each educational grade.

Older students consistently exhibit higher prevalence levels than younger students in the selected baseline and final years.

**Research Conclusion**

Historical cannabis prevalence varies considerably across educational grades and time periods.

Long-term changes should be interpreted within the context of survey coverage, methodological consistency, and historical variability.

The observed differences are descriptive and do not establish statistically significant changes without appropriate survey uncertainty estimates.

```python
import pandas as pd
from pathlib import Path

# Load processed data
DATA_PATH = Path("../data/processed/cannabis_prevalence_clean.csv")
df = pd.read_csv(DATA_PATH)

# Define analysis parameters
GRADES = ["8th Grade", "10th Grade", "12th Grade"]

# Filter past-year cannabis prevalence
cannabis_data = df[
    (df["drug"] == "Cannabis")
    & (df["time"] == "12 Month")
    & (df["grade"].isin(GRADES))
].copy()

# Calculate historical peaks
historical_peaks = (
    cannabis_data.loc[
        cannabis_data.groupby("grade")["estimate"].idxmax()
    ][["grade", "year", "estimate"]]
    .set_index("grade")
    .reindex(GRADES)
    .reset_index()
)

historical_peaks.columns = [
    "Grade",
    "Peak Year",
    "Peak Prevalence (%)"
]

# Compare baseline and final year
comparison = (
    cannabis_data[
        cannabis_data["year"].isin([1991, 2025])
    ]
    .pivot(
        index="grade",
        columns="year",
        values="estimate"
    )
    .reindex(GRADES)
)

comparison["Change (pp)"] = (
    comparison[2025] - comparison[1991]
)

comparison = comparison.rename(
    columns={
        1991: "1991 (%)",
        2025: "2025 (%)"
    }
)

# Display results
print("FINDING 1 | HISTORICAL PREVALENCE")
print("-" * 55)

print("\nHistorical Peaks")
display(historical_peaks)

print("\n1991–2025 Comparison")
display(comparison.round(2))
```

### 3.2 Finding 2 — Recent Cannabis Prevalence Trends (2021–2025)

**Research Question**

How did past-year cannabis prevalence change among 8th, 10th, and 12th-grade students between 2021 and 2025?

**Key Findings**

| Educational Grade | 2021 | 2025 | Change |
|---|---:|---:|---:|
| 8th Grade | 7.1% | 7.6% | +0.5 pp |
| 10th Grade | 17.3% | 15.6% | −1.7 pp |
| 12th Grade | 30.5% | 25.7% | −4.8 pp |

**Interpretation**

Between 2021 and 2025, past-year cannabis prevalence decreased among 10th and 12th-grade students.

The largest absolute decline occurred among 12th-grade students, with a reduction of 4.8 percentage points.

Among 8th-grade students, prevalence increased slightly by 0.5 percentage points.

These results demonstrate that recent prevalence changes differed across educational grades.

**Linear Trend Analysis**

The previously calculated linear regression slopes were:

| Educational Grade | Slope (pp/year) | R² |
|---|---:|---:|
| 8th Grade | +0.18 | 0.139 |
| 10th Grade | −0.70 | 0.495 |
| 12th Grade | −1.45 | 0.873 |

The strongest fitted downward trend occurred among 12th-grade students.

However, the regressions describe only five annual observations and should not be interpreted as evidence of persistent future declines.

**Research Conclusion**

The 2021–2025 period reveals declining past-year cannabis prevalence among older students, particularly those in 12th grade.

The results indicate substantial differences in recent prevalence patterns across educational grades.

These changes remain descriptive and should be interpreted considering the study's methodological limitations.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_PATH = Path("../data/processed/cannabis_prevalence_clean.csv")

df = pd.read_csv(DATA_PATH)

GRADES = ["8th Grade", "10th Grade", "12th Grade"]

cannabis_data = df[
    (df["drug"] == "Cannabis")
    & (df["time"] == "12 Month")
    & (df["grade"].isin(GRADES))
    & (df["year"].between(2021, 2025))
].copy()

results = []

for grade in GRADES:

    grade_data = cannabis_data[
        cannabis_data["grade"] == grade
    ].sort_values("year")

    assert grade_data["year"].tolist() == list(range(2021, 2026)), (
        f"Incomplete annual data for {grade}."
    )

    x = grade_data["year"].to_numpy()
    y = grade_data["estimate"].to_numpy()

    # Calculate linear trend
    slope, intercept = np.polyfit(x - 2021, y, 1)

    predicted = slope * (x - 2021) + intercept

    ss_res = np.sum((y - predicted) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)

    r_squared = 1 - (ss_res / ss_tot) if ss_tot > 0 else np.nan

    results.append({
        "Grade": grade,
        "2021 (%)": y[0],
        "2025 (%)": y[-1],
        "Change (pp)": y[-1] - y[0],
        "Slope (pp/year)": slope,
        "R²": r_squared
    })

findings = pd.DataFrame(results)

print("FINDING 2 | RECENT CANNABIS PREVALENCE TRENDS")
print("-" * 55)

display(findings.round(3))
```

### 3.3 Finding 3 — Cannabis Vaping Trends (2021–2025)

**Research Question**

How did past-year cannabis vaping prevalence change between 2021 and 2025, and how did these changes compare with general cannabis prevalence?

**Key Findings**

| Educational Grade | 2021 | 2025 | Change |
|---|---:|---:|---:|
| 8th Grade | 4.7% | 4.7% | 0.0 pp |
| 10th Grade | 12.4% | 10.6% | −1.8 pp |
| 12th Grade | 18.3% | 16.0% | −2.3 pp |

**Comparison With General Cannabis Prevalence**

| Educational Grade | Cannabis Change | Cannabis Vaping Change |
|---|---:|---:|
| 8th Grade | +0.5 pp | 0.0 pp |
| 10th Grade | −1.7 pp | −1.8 pp |
| 12th Grade | −4.8 pp | −2.3 pp |

**Interpretation**

Between 2021 and 2025, cannabis vaping prevalence decreased among 10th and 12th-grade students.

Among 8th-grade students, the 2021 and 2025 prevalence estimates were identical.

However, the absence of a net change does not imply that prevalence remained constant throughout the observation period.

Cannabis vaping reached higher levels in intermediate years before declining toward 2025.

The results suggest that general cannabis prevalence and cannabis vaping followed partially different trajectories across educational grades.

**Research Conclusion**

Cannabis vaping experienced decreases among older students between 2021 and 2025.

These decreases were not identical to those observed for general cannabis prevalence.

The findings demonstrate the importance of examining cannabis-related product categories separately rather than treating them as interchangeable indicators.

Because cannabis vaping is included within broader cannabis-use measures, these prevalence categories should not be added together.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/processed/cannabis_prevalence_clean.csv")
df = pd.read_csv(DATA_PATH)

GRADES = ["8th Grade", "10th Grade", "12th Grade"]
CATEGORIES = ["Cannabis", "Vaping Cannabis"]

comparison_data = df[
    (df["drug"].isin(CATEGORIES))
    & (df["time"] == "12 Month")
    & (df["grade"].isin(GRADES))
    & (df["year"].isin([2021, 2025]))
].copy()

# Validate expected observations
assert len(comparison_data) == 12, (
    "Unexpected number of observations."
)

assert not comparison_data.duplicated(
    subset=["drug", "grade", "year"]
).any(), "Duplicate observations detected."

# Calculate category-specific changes
comparison = comparison_data.pivot(
    index=["grade", "drug"],
    columns="year",
    values="estimate"
)

assert comparison[[2021, 2025]].notna().all().all(), (
    "Missing baseline or final observations."
)

comparison["Change (pp)"] = comparison[2025] - comparison[2021]

comparison = comparison.rename(
    columns={
        2021: "2021 (%)",
        2025: "2025 (%)"
    }
)

print("FINDING 3 | CANNABIS VS. CANNABIS VAPING")
print("-" * 55)

display(comparison.round(2))

# Compare changes across categories
change_matrix = (
    comparison["Change (pp)"]
    .unstack("drug")
    .reindex(index=GRADES, columns=CATEGORIES)
)

print("\nCATEGORY CHANGE COMPARISON (pp)")
display(change_matrix.round(2))
```

### 3.4 Finding 4 — Flavored Cannabis Vaping Trends (2021–2025)

**Research Question**

How did past-year flavored cannabis vaping prevalence change between 2021 and 2025, and how did its trajectory differ from general cannabis vaping?

**Key Findings**

| Educational Grade | 2021 | 2025 | Change |

| 10th Grade | 5.2% | 5.8% | +0.6 pp |
| 12th Grade | 6.9% | 8.2% | +1.3 pp |

**Comparison With General Cannabis Vaping**

| 10th Grade | −1.8 pp | +0.6 pp |
| 12th Grade | −2.3 pp | +1.3 pp |

**Interpretation**

Flavored cannabis vaping exhibited positive net changes across all three educational grades between 2021 and 2025.

The largest absolute increase occurred among 12th-grade students, at 1.3 percentage points.

These increases contrast with the declines observed in general cannabis vaping among 10th and 12th-grade students.

However, the annual trajectories were not consistently upward.

Flavored cannabis vaping reached higher prevalence levels during intermediate years before declining toward 2025.

**Research Conclusion**

Flavored cannabis vaping followed a different net-change pattern from general cannabis vaping during the observation period.

This finding highlights the importance of analyzing specific product characteristics separately.

The results do not establish changes in individual product preferences or substitution between consumption methods.

Because flavored cannabis vaping is a subset of broader cannabis vaping, the two prevalence categories should not be treated as independent populations.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/processed/cannabis_prevalence_clean.csv")
df = pd.read_csv(DATA_PATH)

GRADES = ["8th Grade", "10th Grade", "12th Grade"]

CATEGORIES = [
    "Vaping Cannabis",
    "Vaping Flavored Cannabis"
]

comparison_data = df[
    (df["drug"].isin(CATEGORIES))
    & (df["time"] == "12 Month")
    & (df["grade"].isin(GRADES))
    & (df["year"].isin([2021, 2025]))
].copy()

assert len(comparison_data) == 12, (
    "Unexpected number of observations."
)

comparison = comparison_data.pivot(
    index=["grade", "drug"],
    columns="year",
    values="estimate"
)

assert comparison[[2021, 2025]].notna().all().all(), (
    "Missing baseline or final observations."
)

comparison["Change (pp)"] = (
    comparison[2025] - comparison[2021]
)

comparison = comparison.rename(
    columns={
        2021: "2021 (%)",
        2025: "2025 (%)"
    }
)

print("FINDING 4 | FLAVORED CANNABIS VAPING")
print("-" * 55)

display(comparison.round(2))

change_matrix = (
    comparison["Change (pp)"]
    .unstack("drug")
    .reindex(index=GRADES, columns=CATEGORIES)
)

print("\nVAPING CATEGORY CHANGE COMPARISON (pp)")
display(change_matrix.round(2))
```

### 3.5 Finding 5 — Cross-Category Statistical Relationships

**Research Question**

How do statistical relationships between cannabis-related prevalence indicators differ across educational grades and analytical approaches?

**Key Findings**

Pearson correlation coefficients were calculated for three cannabis-related categories over the 2021–2025 period.

Two approaches were evaluated:

1. Correlations between annual prevalence levels.
2. Correlations between year-over-year prevalence changes.

#### Correlation Between Prevalence Levels

| Educational Grade | Cannabis vs. Vaping Cannabis | Cannabis vs. Flavored Vaping | Vaping Cannabis vs. Flavored Vaping |
|---|---:|---:|---:|
| 8th Grade | +0.658 | +0.661 | +0.885 |
| 10th Grade | +0.986 | +0.554 | +0.611 |
| 12th Grade | +0.809 | −0.442 | +0.118 |

The strongest observed correlation occurred between general cannabis prevalence and cannabis vaping among 10th-grade students (r = 0.986).

#### Correlation Between Annual Changes

| Educational Grade | Cannabis vs. Vaping Cannabis | Cannabis vs. Flavored Vaping | Vaping Cannabis vs. Flavored Vaping |
|---|---:|---:|---:|
| 8th Grade | +0.562 | +0.323 | +0.850 |
| 10th Grade | +0.968 | +0.890 | +0.895 |
| 12th Grade | +0.644 | +0.029 | +0.599 |

The results demonstrate that correlation coefficients change when prevalence levels are replaced with annual differences.

**Interpretation**

The strength and direction of statistical associations varied across educational grades.

In particular, the association between general cannabis and flavored cannabis vaping among 12th-grade students changed from a negative prevalence-level correlation (r = −0.442) to a near-zero annual-change correlation (r = 0.029).

This illustrates the importance of distinguishing relationships between observed levels from relationships between annual changes.

However, the results should not be interpreted as evidence of individual-level behavioral relationships.

**Statistical Limitations**

- Prevalence-level correlations use five annual observations.
- Year-over-year correlations use only four annual changes.
- Short time series make correlation coefficients unstable.
- Shared temporal trends may influence observed associations.
- Product categories represent overlapping populations.
- Formal statistical significance was not established.

**Research Conclusion**

Statistical relationships between cannabis-related prevalence indicators vary by educational grade and analytical approach.

These findings provide exploratory context but should not be treated as primary evidence of behavioral relationships, causality, or substitution between products.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/processed/cannabis_prevalence_clean.csv")
df = pd.read_csv(DATA_PATH)

GRADES = ["8th Grade", "10th Grade", "12th Grade"]

CATEGORIES = [
    "Cannabis",
    "Vaping Cannabis",
    "Vaping Flavored Cannabis"
]

analysis_data = df[
    (df["drug"].isin(CATEGORIES))
    & (df["time"] == "12 Month")
    & (df["grade"].isin(GRADES))
    & (df["year"].between(2021, 2025))
].copy()

results = []

for grade in GRADES:

    grade_data = analysis_data[
        analysis_data["grade"] == grade
    ]

    prevalence = grade_data.pivot(
        index="year",
        columns="drug",
        values="estimate"
    ).reindex(
        index=range(2021, 2026),
        columns=CATEGORIES
    )

    assert not prevalence.isna().any().any(), (
        f"Incomplete observations for {grade}."
    )

    annual_changes = prevalence.diff().dropna()

    level_corr = prevalence.corr()
    yoy_corr = annual_changes.corr()

    category_pairs = [
        ("Cannabis", "Vaping Cannabis"),
        ("Cannabis", "Vaping Flavored Cannabis"),
        ("Vaping Cannabis", "Vaping Flavored Cannabis")
    ]

    for category_a, category_b in category_pairs:

        results.append({
            "Grade": grade,
            "Category A": category_a,
            "Category B": category_b,
            "Prevalence Correlation": level_corr.loc[
                category_a, category_b
            ],
            "YoY Correlation": yoy_corr.loc[
                category_a, category_b
            ]
        })

correlation_summary = pd.DataFrame(results)

print("FINDING 5 | CROSS-CATEGORY CORRELATIONS")
print("-" * 55)

display(correlation_summary.round(3))
```

## 4. Final Research Conclusions

### 4.1 Summary of Principal Findings

This research examined cannabis-related prevalence among U.S. students in 8th, 10th, and 12th grades using published Monitoring the Future survey estimates.

The analysis covered historical observations extending from 1975 to 2025, with particular emphasis on comparable product-category trends between 2021 and 2025.

Five principal findings emerged.

**1. Historical cannabis prevalence varies considerably across educational grades.**

The highest observed past-year cannabis prevalence estimates were 18.3% among 8th-grade students in 1996, 34.8% among 10th-grade students in 1997, and 50.8% among 12th-grade students in 1979.

These historical peaks demonstrate that cannabis prevalence has experienced substantial fluctuations rather than following a consistently increasing or decreasing trajectory.

**2. Recent cannabis prevalence declined among older students.**

Between 2021 and 2025, past-year cannabis prevalence decreased by 1.7 percentage points among 10th-grade students and 4.8 percentage points among 12th-grade students.

Among 8th-grade students, prevalence increased by 0.5 percentage points.

**3. Cannabis vaping followed partially different patterns from general cannabis use.**

Between 2021 and 2025, past-year cannabis vaping prevalence declined by 1.8 percentage points among 10th-grade students and 2.3 percentage points among 12th-grade students.

Among 8th-grade students, the initial and final prevalence estimates were identical.

**4. Flavored cannabis vaping exhibited positive net changes across all three grades.**

Between 2021 and 2025, past-year flavored cannabis vaping prevalence increased by 0.4 percentage points among 8th-grade students, 0.6 percentage points among 10th-grade students, and 1.3 percentage points among 12th-grade students.

These increases occurred despite intermediate fluctuations and declines from some earlier peaks.

**5. Statistical associations depended on educational grade and analytical approach.**

Pearson correlation coefficients varied across grades and changed when comparing prevalence levels with year-over-year differences.

The results illustrate the importance of evaluating both the statistical method and the characteristics of the available observations.

Because these correlations were calculated using short time series, they remain exploratory.

### 4.2 Overall Interpretation

The findings indicate that cannabis-related prevalence among U.S. adolescents cannot be adequately described by a single historical trend.

Patterns differ across educational grades, product categories, reporting intervals, and observation periods.

In particular, the contrast between general cannabis vaping and flavored cannabis vaping demonstrates the analytical value of examining specific product characteristics.

However, the results describe aggregate prevalence estimates rather than individual-level behavioral transitions.

They do not demonstrate changes in product preferences, substitution between consumption methods, or causal mechanisms.

### 4.3 Research Limitations

Several methodological limitations must be considered:

**Survey Representativeness**

Monitoring the Future is a school-based survey. Its estimates may not represent adolescents who are absent from school or outside the school system.

**Historical Coverage**

The 12th-grade series begins in 1975, while 8th- and 10th-grade observations begin in 1991.

Historical comparisons across all three grades are therefore restricted to periods with overlapping data coverage.

**Survey Methodology**

Survey disruptions in 2020, changes in administration during 2021, and methodological updates in 2024 may affect historical comparability.

**Aggregate Data**

The dataset contains published prevalence estimates rather than individual-level observations.

Individual behavioral transitions and relationships cannot be identified.

**Statistical Uncertainty**

The dataset does not include the information required to calculate design-based confidence intervals or formal significance tests.

**Short Time Series**

Regression and correlation analyses covering 2021–2025 use only five annual observations.

Correlations between annual changes use only four observations.

These results are descriptive and exploratory rather than confirmatory.

**Overlapping Categories**

Cannabis-related product indicators may represent overlapping populations and should not be added together.

### 4.4 Opportunities for Future Research

Future investigations could extend this research by:

1. Incorporating published confidence intervals or survey design information.
2. Examining longer comparable periods for individual product categories.
3. Investigating the effects of questionnaire and sampling changes.

5. Evaluating differences between cannabis-related consumption measures using harmonized survey definitions.

### 4.5 Final Research Statement

The analysis demonstrates the value of combining data validation, exploratory analysis, descriptive statistics, and transparent methodological documentation.

Rather than establishing causal explanations, the research provides a reproducible description of cannabis-related prevalence trends among U.S. adolescents.

The principal contribution is the identification and comparison of historical patterns across educational grades and cannabis product categories, supported by documented analytical procedures and explicit limitations.
